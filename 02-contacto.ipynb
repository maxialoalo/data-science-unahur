````python
import pandas as pd
````

````python
datos = {
    "anime": [
        "Bleach", "One Piece", "Sword Art Online", "Made in Abyss",
        "Dragon Ball", "Parasyte", "Smoking Behind the Supermarket with You",
        "Black Clover", "Fullmetal Alchemist", "Black Torch",
        "KonoSuba", "Love Live!", "Naruto", "Death Note",
        "Attack on Titan", "Hunter x Hunter", "Demon Slayer",
        "Jujutsu Kaisen", "Steins;Gate", "Mob Psycho 100"
    ],

    "genero": [
        "Accion", "Aventura", "Isekai", "Aventura",
        "Accion", "Terror", "Romance", "Fantasia",
        "Aventura", "Accion", "Comedia", "Musical",
        "Accion", "Suspenso", "Accion", "Aventura",
        "Accion", "Accion", "Ciencia ficcion", "Comedia"
    ],

    "episodios": [
        366, 1180, 96, 25, 153, 24, 12, 170, 51, 12,
        31, 13, 220, 37, 94, 148, 63, 59, 24, 37
    ],

    "anio_inicio": [
        2004, 1999, 2012, 2017, 1986, 2014, 2023, 2017,
        2003, 2024, 2016, 2013, 2002, 2006, 2013, 2011,
        2019, 2020, 2011, 2016
    ],

    "calificacion": [
        8.6, 9.0, 7.8, 8.5, 8.7, 8.3, 7.5, 8.2,
        8.9, 7.0, 8.1, 7.8, 8.5, 9.0, 9.5, 9.0,
        8.7, 8.6, 9.1, 8.9
    ]
}
````

````python
df=pd.DataFrame(datos)
df
````

````python
El dataset contiene 20 observaciones, donde cada observación representa un anime. 
    Además cuenta con 5 variables que permiten describir caracteristicas de cada uno
````

````python
Cual es el promedio de calificacion de los animes?
````

````python
promedio_calificacion=df["calificacion"].mean()
promedio_calificacion
````

````python
La calificación promedio de los animes registrados es de aproximadamente 8.48 puntos.
Esto indica que dentro del dataset las calificaciones son relativamente altas.
````

````python
Los animes más antiguos tienen más episodios que los recientes?
````

````python
df[["anime","anio_inicio","episodios"]].sort_values(by="anio_inicio")
````

````python
Al observar los años de inicio y la cantidad de episodios, se puede identificar una tendencia: 
varios de los animes más antiguos tienen una cantidad considerablemente mayor de episodios que los más recientes. 
Esto podría indicar que, en este conjunto de datos, las series antiguas tienden a tener una mayor duración. 
Sin embargo, existen excepciones y no significa que todos los animes antiguos tengan más episodios que los nuevos.
````

````python
Los animes con más episodios tienen mejores calificaciones?
````

````python
df2 = df[["anime", "episodios", "calificacion"]].sort_values(by="calificacion", ascending=False)
df2
````

````python
Los géneros influyen en la cantidad de episodios?
````

````python
df2=df[["anime","genero","episodios"]].sort_values(by="genero")
df2
````

| Herramienta | Uso |
|---|---|
| `import pandas as pd` | Importa la biblioteca Pandas y permite utilizarla mediante el alias `pd`. |
| `pd.DataFrame()` | Convierte un diccionario u otros datos en un DataFrame (tabla). |
| `head()` | Muestra por defecto las primeras 5 filas del DataFrame. |
| `shape` | Devuelve la cantidad de filas y columnas del DataFrame. |
| `info()` | Muestra información general del DataFrame, como columnas, cantidad de valores no nulos y tipos de datos. |
| `describe()` | Genera estadísticas descriptivas de las columnas numéricas, como promedio, mínimo, máximo y cuartiles. |
| `len()` | Devuelve la cantidad de elementos, en este caso, la cantidad de filas del DataFrame. |
| `mean()` | Calcula el promedio de los valores de una columna numérica. |
| `sort_values()` | Ordena los datos según los valores de una o varias columnas. |
| `ascending=False` | Permite ordenar los valores de mayor a menor. |
| `df["columna"]` | Permite seleccionar una columna específica del DataFrame. |
| `df[["columna1", "columna2"]]` | Permite seleccionar varias columnas del DataFrame. |
